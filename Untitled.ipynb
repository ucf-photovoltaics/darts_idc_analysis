```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt
import re

# -------------------------
# Helper: Clean CSV reader
# -------------------------
def clean_csv(path):
    df = pd.read_csv(path, on_bad_lines="skip")
    for col in ["Time (ms)", "Current (mA)"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")
    df = df.dropna(subset=["Time (ms)", "Current (mA)"])
    return df

# -------------------------
# Config
# -------------------------
folder = r"C:\Users\goodn\OneDrive\Documents\GitHub\darts_idc_analysis\Data\CurrentTime"
all_data = []

# -------------------------
# Load + Clean All CSVs
# -------------------------
for file_path in glob.glob(os.path.join(folder, "*.csv")):
    try:
        df = clean_csv(file_path)

        fname = os.path.basename(file_path)

        # Extract board type (01, 04, 07, 10) from filename
        match = re.search(r"(01|04|07|10)", fname)
        if match:
            board_type = match.group(1)
        else:
            board_type = "Unknown"

        df["Board_Type"] = board_type
        df["Sensor_ID"] = fname.replace(".csv", "")
        df["Time_to_Failure"] = df["Time (ms)"] - df["Time (ms)"].min()

        all_data.append(df)
    except Exception as e:
        print(f"Skipping {file_path}: {e}")

# -------------------------
# Combine all data
# -------------------------
if all_data:
    data = pd.concat(all_data, ignore_index=True)
else:
    raise ValueError("No valid CSVs found!")

# -------------------------
# Plot: Separate per Board Type
# -------------------------
board_types = data["Board_Type"].unique()
n = len(board_types)

fig, axes = plt.subplots(n, 1, figsize=(10, 5*n), sharex=True)

if n == 1:
    axes = [axes]  # Ensure iterable if only one board type

for ax, board in zip(axes, board_types):
    subset = data[data["Board_Type"] == board]
    for sid, df_sensor in subset.groupby("Sensor_ID"):
        ax.plot(df_sensor["Time_to_Failure"], df_sensor["Current (mA)"], alpha=0.6, label=sid)

    ax.set_title(f"Board Type {board} — Current vs Time to Failure")
    ax.set_ylabel("Current (mA)")
    ax.grid(True)


axes[-1].set_xlabel("Time to Failure (ms)")
plt.tight_layout()
plt.show()
```
