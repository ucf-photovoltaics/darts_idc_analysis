```python
#---------------setup and imports ---------

import sys
sys.path.append("C:/Users/goodn/OneDrive/Documents/GitHub/darts_idc_analysis/Analysis")

import adds
import pandas as pd
import seaborn as sns

%matplotlib inline
import matplotlib.pyplot as plt

#------Loading Data----------
master_current_time = adds.get_master_current_time()
print(master_current_time["Solution"].unique())
print(master_current_time["Board ID"].unique())

#--------Plotting Loop ------
solutions = master_current_time["Solution"].unique()
boards = master_current_time["Board ID"].unique()

for sol in solutions:
    for board in boards:
        df = master_current_time[
            (master_current_time["Solution"] == sol) &
            (master_current_time["Board ID"] == board)
        ]
        if df.empty:
            continue

        plt.figure(figsize=(8, 4))
        sns.lineplot(data=df, x="Time (ms)", y="Current (mA)", hue="Sensor", marker="o")
        plt.title(f"Current vs Time - Board {board}, Solution {sol}")
        plt.xlabel("Time (ms)")
        plt.ylabel("Current (mA)")
        plt.legend(title="Sensor")
        plt.tight_layout()
        plt.show()
```
