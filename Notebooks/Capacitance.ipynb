```python
#-----Setup and imports--------------

import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline


#---------load and combine data-----------
base_path = "C:/Users/goodn/OneDrive/Documents/GitHub/darts_idc_analysis"
data_types = ["CV", "CF"]
conditions = ["_PRISTINE", "_EXPOSED"]

all_data = []

for dtype in data_types:
    for cond in conditions:
        folder = os.path.join(base_path, f"{dtype}{cond}")
        for file in os.listdir(folder):
            if file.endswith(".csv"):
                df = pd.read_csv(os.path.join(folder, file))
                df["Data Type"] = dtype
                df["Condition"] = cond.strip("_")
                all_data.append(df)

cap_data = pd.concat(all_data, ignore_index=True)
print(cap_data.head())


#------plotting------- 
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=cap_data,
    x="Frequency (Hz)",        # replace with actual column
    y="Capacitance (F)",       # replace with actual column
    hue="Condition",           # Pristine vs Exposed
    style="Data Type"          # CV vs CF
)
plt.title("Capacitance vs Frequency")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Capacitance (F)")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()


```
