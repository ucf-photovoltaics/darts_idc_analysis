```python
import os
import cv2
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Paths
pristine_path = r"C:\Users\goodn\OneDrive\Documents\GitHub\darts_idc_analysis\Data\Imgscans_PRISTINE_sensors"
exposed_path = r"C:\Users\goodn\OneDrive\Documents\GitHub\darts_idc_analysis\Data\Imgscans_EXPOSED_sensors"
figures_path = r"C:\Users\goodn\OneDrive\Documents\GitHub\darts_idc_analysis\results\figures_rgb"
os.makedirs(figures_path, exist_ok=True)

# Function to read image and compute mean R,G,B, Grayscale
def average_rgb(image_path):
    img = cv2.imread(image_path)
    if img is None:
        return None
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    mean_rgb = img_rgb.mean(axis=(0,1))
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).mean()
    return {"R": mean_rgb[0], "G": mean_rgb[1], "B": mean_rgb[2], "Grayscale": gray}

# Collect data
board_data = []

# Pristine
for f in os.listdir(pristine_path):
    if f.endswith(".jpg"):
        parts = f.split("_")
        board_type = parts[1]
        sensor_id = parts[-1].split(".")[0]
        vals = average_rgb(os.path.join(pristine_path, f))
        if vals:
            vals.update({
                "Board_Type": board_type,
                "Sensor_ID": sensor_id,
                "Condition": "Pristine",
                "Filename": f
            })
            board_data.append(vals)

# Exposed
for f in os.listdir(exposed_path):
    if f.endswith(".jpg"):
        parts = f.split("_")
        board_type = parts[1]
        sensor_id = parts[-1].split(".")[0]
        vals = average_rgb(os.path.join(exposed_path, f))
        if vals:
            vals.update({
                "Board_Type": board_type,
                "Sensor_ID": sensor_id,
                "Condition": "Exposed",
                "Filename": f
            })
            board_data.append(vals)

df = pd.DataFrame(board_data)

# Compute differences
diff_data = []
for board_type in df["Board_Type"].unique():
    for sensor in df["Sensor_ID"].unique():
        pristine_vals = df[(df["Board_Type"]==board_type) &
                           (df["Sensor_ID"]==sensor) &
                           (df["Condition"]=="Pristine")]
        if pristine_vals.empty:
            continue
        exposed_vals = df[(df["Board_Type"]==board_type) &
                          (df["Sensor_ID"]==sensor) &
                          (df["Condition"]=="Exposed")]
        for _, row in exposed_vals.iterrows():
            diff_data.append({
                "Board_Type": board_type,
                "Sensor_ID": sensor,
                "Red_Diff": row["R"] - pristine_vals["R"].values[0],
                "Green_Diff": row["G"] - pristine_vals["G"].values[0],
                "Blue_Diff": row["B"] - pristine_vals["B"].values[0],
                "Gray_Diff": row["Grayscale"] - pristine_vals["Grayscale"].values[0]
            })

diff_df = pd.DataFrame(diff_data)

# Save summary CSV
csv_path = os.path.join(figures_path, "image_differences_summary.csv")
diff_df.to_csv(csv_path, index=False)
print("✅ Image differences summary saved at:", csv_path)

# Melt for plotting
diff_melted = diff_df.melt(
    id_vars=["Board_Type","Sensor_ID"],
    value_vars=["Red_Diff","Green_Diff","Blue_Diff","Gray_Diff"],
    var_name="Channel",
    value_name="Difference"
)

# Palette
palette = {"Red_Diff":"#FF0000","Green_Diff":"#00FF00","Blue_Diff":"#0000FF","Gray_Diff":"#888888"}

# Boxplots per board
for board in diff_melted["Board_Type"].unique():
    plt.figure(figsize=(10,6))
    board_df = diff_melted[diff_melted["Board_Type"]==board]
    sns.boxplot(
        data=board_df,
        x="Sensor_ID",
        y="Difference",
        hue="Channel",
        palette=palette
    )
    plt.title(f"RGB / Grayscale Difference vs Pristine - Board {board}")
    plt.ylabel("Difference")
    plt.xlabel("Sensor")
    plt.legend(title="Channel", bbox_to_anchor=(1.05,1), loc='upper left')
    plt.tight_layout()
    plt.savefig(os.path.join(figures_path, f"RGB_Gray_Boxplot_Board_{board}.png"), dpi=300)
    plt.show()

# Faceted scatterplots per channel
g = sns.FacetGrid(diff_melted, col="Channel", col_wrap=2, height=4, sharey=True)
g.map_dataframe(sns.scatterplot, x="Sensor_ID", y="Difference", hue="Board_Type", alpha=0.7)
g.fig.suptitle("ΔColor / Gray vs Sensor_ID (All Boards)", y=1.02)
plt.tight_layout()
plt.legend(title="Board", bbox_to_anchor=(1.05,1), loc='upper left')
plt.savefig(os.path.join(figures_path, "scatter_diff_per_channel.png"), dpi=300)
plt.show()

# Faceted scatterplots per board
g2 = sns.FacetGrid(diff_melted, col="Board_Type", col_wrap=2, height=4, sharey=True)
g2.map_dataframe(sns.scatterplot, x="Sensor_ID", y="Difference", hue="Channel", palette=palette, alpha=0.7)
g2.fig.suptitle("ΔColor / Gray vs Sensor_ID (Per Board)", y=1.02)
plt.tight_layout()
plt.legend(title="Intensity", bbox_to_anchor=(1.05,1), loc='upper left')

plt.savefig(os.path.join(figures_path, "scatter_diff_per_board.png"), dpi=300)
plt.show()

print("🎨 Plots complete. Figures saved in:", figures_path)
```
