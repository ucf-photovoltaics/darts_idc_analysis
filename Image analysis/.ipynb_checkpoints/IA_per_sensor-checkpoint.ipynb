```python
import sys
sys.path.append(r"C:\Users\goodn\OneDrive\Documents\GitHub\darts_idc_analysis\Analysis")
import adds
import seaborn as sns
import matplotlib.pyplot as plt
import os
import pandas as pd
import numpy as np
import cv2

pristine_path = r"C:\Users\goodn\OneDrive\Documents\GitHub\darts_idc_analysis\data\Imgscans_PRISTINE_sensors"
exposed_path = r"C:\Users\goodn\OneDrive\Documents\GitHub\darts_idc_analysis\data\Imgscans_EXPOSED_sensors"
figures_path = r"C:\Users\goodn\OneDrive\Documents\GitHub\darts_idc_analysis\results\figures_rgb"
os.makedirs(figures_path, exist_ok=True)

# -------------------------
# Helper functions
# -------------------------
def average_rgb(image_path):
    img = cv2.imread(image_path)
    if img is None:
        return None
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    mean_rgb = img.mean(axis=(0,1))
    gray = 0.2989*img[:,:,0] + 0.5870*img[:,:,1] + 0.1140*img[:,:,2]
    return {
        "R": mean_rgb[0],
        "G": mean_rgb[1],
        "B": mean_rgb[2],
        "Brightness": gray.mean()
    }

# -------------------------
# Load pristine images
# -------------------------
pristine_files = [f for f in os.listdir(pristine_path) if f.endswith(".jpg")]
pristine_data = []
for f in pristine_files:
    parts = f.split("_")
    board_type = parts[1]
    sensor_id = parts[-1].split(".")[0]
    vals = average_rgb(os.path.join(pristine_path, f))
    if vals:
        vals.update({
            "Board_Type": board_type,
            "Sensor_ID": sensor_id,
            "Condition": "PRISTINE",
            "Filename": f
        })
        pristine_data.append(vals)
pristine_df = pd.DataFrame(pristine_data)

# -------------------------
# Load exposed images
# -------------------------
exposed_files = [f for f in os.listdir(exposed_path) if f.endswith(".jpg")]
exposed_data = []
for f in exposed_files:
    parts = f.split("_")
    board_type = parts[1]
    sensor_id = parts[-1].split(".")[0]
    vals = average_rgb(os.path.join(exposed_path, f))
    if vals:
        vals.update({
            "Board_Type": board_type,
            "Sensor_ID": sensor_id,
            "Condition": "EXPOSED",
            "Filename": f
        })
        exposed_data.append(vals)
exposed_df = pd.DataFrame(exposed_data)

# -------------------------
# Compute differences
# -------------------------
diff_df = exposed_df.merge(
    pristine_df,
    on=["Board_Type","Sensor_ID"],
    suffixes=("_EXPOSED","_PRISTINE")
)

diff_df["Red_Diff"] = diff_df["R_EXPOSED"] - diff_df["R_PRISTINE"]
diff_df["Green_Diff"] = diff_df["G_EXPOSED"] - diff_df["G_PRISTINE"]
diff_df["Blue_Diff"] = diff_df["B_EXPOSED"] - diff_df["B_PRISTINE"]
diff_df["Brightness_Diff"] = diff_df["Brightness_EXPOSED"] - diff_df["Brightness_PRISTINE"]

# -------------------------
# Melt for combined RGB + Grayscale
# -------------------------
combined_melt = pd.melt(
    diff_df,
    id_vars=["Board_Type","Sensor_ID"],
    value_vars=["Red_Diff","Green_Diff","Blue_Diff","Brightness_Diff"],
    var_name="Channel",
    value_name="Difference"
)

combined_melt["Channel"] = combined_melt["Channel"].replace({
    "Red_Diff":"Red",
    "Green_Diff":"Green",
    "Blue_Diff":"Blue",
    "Brightness_Diff":"Grayscale"
})

palette = {
    "Red":"#FF0000",
    "Green":"#00FF00",
    "Blue":"#0000FF",
    "Grayscale":"#888888"
}


# -------------------------
# Per-sensor per-board plots (no pristine dots)
# -------------------------
os.makedirs(os.path.join(figures_path,"per_sensor"), exist_ok=True)
board_sensor_combos = combined_melt[["Board_Type","Sensor_ID"]].drop_duplicates()

for _, row in board_sensor_combos.iterrows():
    board = row["Board_Type"]
    sensor = row["Sensor_ID"]

    subset = combined_melt[
        (combined_melt["Board_Type"]==board) &
        (combined_melt["Sensor_ID"]==sensor)
    ]
    
    plt.figure(figsize=(8,5))
    sns.boxplot(
        data=subset,
        x="Channel",
        y="Difference",
        hue="Channel",  # use hue to properly assign palette in v0.14+
        palette=palette,
        dodge=False
    )
    
    plt.title(f"RGB + Grayscale Differences\nBoard {board} - Sensor {sensor}")
    plt.xlabel("Channel")
    plt.ylabel("Difference")
    plt.tight_layout()
    
    save_name = f"Board_{board}_Sensor_{sensor}_RGB_Gray_diff.png"
    plt.savefig(os.path.join(figures_path,"per_sensor",save_name), dpi=300)
    plt.show()
print("saved in", figures_path)
```
